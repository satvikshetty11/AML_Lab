```python
#Create Arrays
import numpy as np

a = np.array([10,20,30,40])
print(a)
```

```python
b = np.arange(1,11)
print(b)
```

```python
c = np.linspace(0,1,5)
print(c)
```

```python
#Matrix operation
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(A+B)
print(A-B)
print(A*B)
print(A@B)
```

```python
#Statistics
marks = np.array([70,65,80,90,88])

print(np.mean(marks))
print(np.min(marks))
print(np.max(marks))
print(np.std(marks))
```

```python
#Temperature
temp = np.array([29, 29, 29, 26, 27, 28, 29, 30, 32, 32, 32, 33, 33, 34, 33, 32, 31, 30, 31, 28, 31, 29, 26, 30, 29, 31, 30, 29, 28, 28, 29])
max_temp = np.max(temp)
min_temp = np.min(temp)
avg_temp = np.mean(temp)
```

```python
hottest_day = np.argmax(temp) + 1
coldest_day = np.argmin(temp) + 1
```

```python
print(f"Hottest Temperature : {max_temp}°C on July {hottest_day}, 2026")
print(f"Coldest Temperature : {min_temp}°C on July {coldest_day}, 2026")
print(f"Average Temperature : {avg_temp:.2f}°C")
```
